```python
import numpy as np
import sklearn
import matplotlib
import matplotlib.pyplot as plt
```

# Simple case

```python
n_prices = 6
np.random.seed(6)
prices = np.random.rand(n_prices)*20
prices.sort()
prices_sup_10 = prices>10
prices
```

```python
plt.scatter(prices,prices_sup_10)
plt.tight_layout()
plt.savefig("linear_classif_noline.pdf")
plt.show()
```

```python
prices[2]-prices[1]
```

```python
prices[-3] - prices[-2]
```

```python
model = sklearn.linear_model.LinearRegression()
model.fit(prices.reshape(n_prices,1), prices_sup_10.reshape(n_prices,1))
pred = model.predict(prices.reshape(n_prices,1))
half_line = ((0.5-model.intercept_)/model.coef_)[0,0]
```

```python
half_line
```

```python
plt.plot(prices, pred, color='red')
plt.scatter(prices,prices_sup_10)
plt.axhline(0.5, color='grey', linestyle="--")
plt.axvline(half_line, color='green', linestyle=":")
plt.text(half_line-4, 1.1, s="decision=no", color="green")
plt.text(half_line+1, 1.1, s="decision=yes", color="green")
plt.tight_layout()
plt.savefig("linear_classif_simple.pdf")
plt.show()
```

# Outlier case

```python
prices = np.append(prices, [100])
prices_sup_10 = prices>10
n_prices = 7
```

# Imbalance

```python
n_prices = 100
prices = np.append(np.random.rand(int(n_prices*.05))*10, np.random.rand(int(n_prices*.95))*10+10)
prices.sort()
prices_sup_10 = prices>10
prices
```

```python
model = sklearn.linear_model.LinearRegression()
model.fit(prices.reshape(n_prices,1), prices_sup_10.reshape(n_prices,1))
pred = model.predict(prices.reshape(n_prices,1))
half_line = ((0.5-model.intercept_)/model.coef_)[0,0]
```

```python
plt.plot(prices, pred, color='red')
plt.scatter(prices,prices_sup_10)
plt.axhline(0.5, color='grey', linestyle="--")
plt.axvline(half_line, color='green', linestyle=":")
plt.text(half_line-5, 1.1, s="decision=no", color="green")
plt.text(half_line+1, 1.1, s="decision=yes", color="green")
plt.tight_layout()
plt.savefig("linear_classif_imbalance.pdf")
plt.show()
```
